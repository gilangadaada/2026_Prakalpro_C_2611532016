# Boolean

Tipe data logika yang hanya memiliki dua nilai, yaitu True (benar) dan False (salah), yang biasanya digunakan untuk mengevaluasi suatu kondisi atau percabangan (seperti menentukan status kelulusan)

# Numerik

Tipe data yang merepresentasikan angka, yang terbagi menjadi bilangan bulat (int), bilangan desimal/pecahan (float), dan bilangan kompleks

# String

Tipe data yang digunakan untuk menyimpan teks atau karakter, baik berupa kata, kalimat, maupun teks multi-baris yang diapit dengan tanda kutip.

# Konstanta

Variabel khusus (menggunakan Final) yang nilainya bersifat tetap dan tidak boleh diubah setelah didefinisikan, contohnya nilai π (pi) untuk perhitungan matematika

Berdasarkan pelaksanaan praktikum pekan kedua mengenai dasar-dasar pemrograman Python, dapat disimpulkan beberapa hal penting sebagai berikut:

* Python menyediakan berbagai macam tipe data numerik seperti *integer*, *float*, dan *complex* yang jenisnya dapat diperiksa menggunakan fungsi `type()`.


* Pengelolaan data teks (string) dapat dilakukan secara fleksibel menggunakan tanda kutip tunggal, ganda, maupun *triple quotes* (`"""`) untuk string multi-baris seperti data alamat.


* Tipe data *boolean* (`True`/`False`) dapat dikombinasikan dengan operator logika (`and`) serta struktur kondisional `if` untuk melakukan evaluasi dan pengambilan keputusan dalam program.


* Fungsi `input()` memungkinkan program menerima data secara interaktif dari pengguna, di mana data tersebut perlu dikonversi (*casting*) menggunakan fungsi seperti `int()` atau `float()` agar dapat dioperasikan secara matematis.


* Konstanta dapat didefinisikan menggunakan modul `typing` (`Final`) untuk memastikan nilai variabel tertentu (seperti nilai konstanta matematika `PI`) bersifat tetap dan tidak berubah selama program berjalan.
