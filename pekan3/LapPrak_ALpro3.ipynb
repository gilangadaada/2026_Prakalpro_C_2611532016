## 2.1 Dasar Teori
### 2.1.1 Pengertian Operator

Operator adalah simbol atau kata khusus yang digunakan untuk melakukan operasi terhadap satu atau lebih operand. Operand dapat berupa nilai, variabel, maupun objek. Dalam Python, operator digunakan untuk melakukan berbagai proses seperti perhitungan, perbandingan, penggabungan kondisi, hingga manipulasi nilai biner.

Beberapa jenis operator yang digunakan dalam praktikum ini antara lain operator aritmatika, assignment, perbandingan, logika, membership, identity, dan bitwise.

### 2.1.2 Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi matematika. Beberapa operator aritmatika yang terdapat pada Python yaitu:

Operator	Fungsi
+	Penjumlahan
-	Pengurangan
*	Perkalian
/	Pembagian
//	Pembagian bulat
%	Sisa pembagian atau modulus
**	Perpangkatan

Dalam program transaksi, operator aritmatika dapat digunakan untuk menghitung besar diskon, total pembayaran, rata-rata harga barang, dan sisa pembagian.

### 2.1.3 Operator Assignment

Operator assignment digunakan untuk memberikan atau memperbarui nilai pada sebuah variabel. Operator dasar assignment adalah =.
Python juga menyediakan augmented assignment, yaitu operator yang menggabungkan operasi aritmatika dengan assignment.

Contohnya:

nilai_2016 = 10
nilai_2016 += 5

Setelah operasi tersebut, nilai nilai_2016 menjadi 15.

Beberapa augmented assignment yang umum digunakan yaitu +=, -=, *=, /=, dan lainnya.

### 2.1.4 Operator Perbandingan

Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil dari operator perbandingan berupa nilai Boolean, yaitu True atau False.

Beberapa operator perbandingan antara lain:

Operator	Fungsi
==	Sama dengan
!=	Tidak sama dengan
>	Lebih besar
<	Lebih kecil
>=	Lebih besar atau sama dengan
<=	Lebih kecil atau sama dengan

Dalam program transaksi, operator perbandingan digunakan untuk menentukan apakah total belanja telah mencapai batas minimum dan apakah jumlah barang telah memenuhi persyaratan.

### 2.1.5 Operator Logika

Operator logika digunakan untuk menggabungkan atau membalik kondisi Boolean. Python memiliki tiga operator logika utama, yaitu and, or, dan not.

and menghasilkan True jika seluruh kondisi bernilai True.
or menghasilkan True jika setidaknya salah satu kondisi bernilai True.
not membalik nilai Boolean.

Contohnya:

member_2016 = True
belanja_2016 = True

hasil_2016 = member_2016 and belanja_2016

Hasil dari operasi tersebut adalah True.

### 2.1.6 Operator Membership

Operator membership digunakan untuk memeriksa apakah suatu nilai terdapat di dalam suatu koleksi data seperti list, tuple, atau string.

Operator membership yang digunakan dalam Python adalah:

in
not in

Contohnya:

kode_2016 = "HEMAT10"
daftar_2016 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

hasil_2016 = kode_2016 in daftar_2016

Hasilnya adalah True karena kode HEMAT10 terdapat di dalam daftar.

### 2.1.7 Operator Identity

Operator identity digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama.

Operator identity terdiri dari:

is
is not

Operator is membandingkan identitas objek, sedangkan == membandingkan nilai atau isi objek.

Sebagai contoh, dua list dapat memiliki isi yang sama tetapi merupakan dua objek yang berbeda. Oleh karena itu, kedua list tersebut dapat menghasilkan True ketika dibandingkan dengan ==, tetapi False ketika dibandingkan menggunakan is.

### 2.1.8 Operator Bitwise

Operator bitwise digunakan untuk melakukan operasi terhadap bilangan dalam representasi biner. Operator ini dapat digunakan untuk menggabungkan maupun memeriksa status tertentu yang direpresentasikan menggunakan bit.

Beberapa operator bitwise yang digunakan dalam praktikum adalah:

Operator	Fungsi
&	AND bitwise
|	OR bitwise
^	XOR bitwise
<<	Left shift
>>	Right shift

Pada studi kasus transaksi, setiap kondisi pelanggan direpresentasikan menggunakan nilai bit:

Nilai Bit	Kondisi
0001	Pelanggan merupakan member
0010	Total belanja ≥ Rp200.000
0100	Jumlah barang ≥ 3
1000	Kode promo tersedia

Operator OR digunakan untuk menggabungkan beberapa kondisi menjadi satu kode status. Operator AND digunakan untuk memeriksa kondisi tertentu dari kode status. Sementara itu, XOR dapat digunakan untuk membandingkan perbedaan antara kode transaksi dan kode referensi.

```python
angka1_2016 = int(input("angka-1: "))
angka2_2016 = int(input("angka-2: "))

#Penjumlahan
hasil_2016 = angka1_2016 + angka2_2016
print("\nOperator penjumlahan")
print("hasil =", hasil_2016)

#Pengurangan
hasil_2016 = angka1_2016 - angka2_2016
print("\nOperator pengurangan")
print("hasil =", hasil_2016)

#Perkalian
hasil_2016 = angka1_2016 * angka2_2016
print("\nOperator perkalian")
print("hasil =", hasil_2016)

#Pembagian, pembagian bulat, dan sisa bagi
if angka2_2016 != 0:
    hasil_2016 = angka1_2016 / angka2_2016
    print("\nOperator pembagian")
    print("hasil =", hasil_2016)

    hasil_2016 = angka1_2016 // angka2_2016
    print("\nOperator pembagian bulat")
    print("hasil =", hasil_2016)

    hasil_2016 = angka1_2016 % angka2_2016
    print("\nOperator sisa bagi")
    print("hasil =", hasil_2016)
else:
    print("Angka kedua tidak boleh bernilai 0.")

#Pangkat
hasil_2016 = angka1_2016 ** angka2_2016
print("\nOperator pangkat")
print("hasil =", hasil_2016)
```

### 2.2.2 Program "assignment_2611532016"

```python
angka1_2016 = int(input("angka-1: "))
angka2_2016 = int(input("angka-2: "))

print("\nNilai awal angka1_2016 =", angka1_2016)
print("Nilai awal angka2_2016 =", angka2_2016)

# Assignment biasa
hasil_2016 = angka1_2016
print("\nAssignment biasa (=)")
print("hasil =", hasil_2016)

# Assignment penambahan
hasil_2016 = angka1_2016
hasil_2016 += angka2_2016
print("\nAssignment penambahan (+=)")
print("hasil =", hasil_2016)

# Assignment pengurangan
hasil_2016 = angka1_2016
hasil_2016 -= angka2_2016
print("\nAssignment pengurangan (-=)")
print("hasil =", hasil_2016)

# Assignment perkalian
hasil_2016 = angka1_2016
hasil_2016 *= angka2_2016
print("\nAssignment perkalian (*=)")
print("hasil =", hasil_2016)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2016 != 0:
    hasil_2016 = angka1_2016
    hasil_2016 /= angka2_2016
    print("\nAssignment pembagian (/=)")
    print("hasil =", hasil_2016)

    #Operator tambahan
    hasil_2016 = angka1_2016
    hasil_2016 //= angka2_2016
    print("\nAssignment pembagian bulat (//=)")
    print("hasil =", hasil_2016)
    hasil_2016 = angka1_2016
    hasil_2016 %= angka2_2016
    print("\nAssignment sisa bagi (%=)")
    print("hasil =", hasil_2016)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_2016 = angka1_2016
hasil_2016 **= angka2_2016
print("\nAssignment perpangkatan (**=)")
print("hasil =", hasil_2016)
```

### 2.2.6 Program "bitwise_2611532016"

```python
print("\n==================================")
print("3. OPERATOR BITWISE")
print("==================================")

angka1_2016 = int(input("Masukkan angka bitwise-1: "))
angka2_2016 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner:")
print("Angka 1:", angka1_2016, "-| biner", bin(angka1_2016))
print("Angka 2:", angka2_2016, "-| biner", bin(angka2_2016))

# Bitwise AND
hasil_2016 = angka1_2016 & angka2_2016
print("\nBitwise AND (&)")
print(angka1_2016, "&", angka2_2016, "=", hasil_2016)
print("Biner hasil =", bin(hasil_2016))
print("Biner hasil (8 bit) =", format(hasil_2016, '08b'))

# Bitwise OR
hasil_2016 = angka1_2016 | angka2_2016
print("\nBitwise OR (|)")
print(angka1_2016, "|", angka2_2016, "=", hasil_2016)
print("Biner hasil =", bin(hasil_2016))
print("Biner hasil (8 bit) =", format(hasil_2016, '08b'))

# Bitwise XOR
hasil_2016 = angka1_2016 ^ angka2_2016
print("\nBitwise XOR (^)")
print(angka1_2016, "^", angka2_2016, "=", hasil_2016)
print("Biner hasil =", bin(hasil_2016))
print("Biner hasil (8 bit) =", format(hasil_2016, '08b'))

# Bitwise NOT
hasil_2016 = ~angka1_2016
print("\nBitwise NOT (~)")
print("~", angka1_2016, "=", hasil_2016)
print("Biner hasil =", bin(hasil_2016))
print("Biner hasil (8 bit) =", format(hasil_2016, '08b'))

# Bitwise geser kiri
jumlah_geser_2016 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2016 = angka1_2016 << jumlah_geser_2016
print("\nBitwise Geser Kiri (<<)")
print(angka1_2016, "<<", jumlah_geser_2016, "=", hasil_2016)
print("Biner hasil =", bin(hasil_2016))
print("Biner hasil (8 bit) =", format(hasil_2016, '08b'))

# Bitwise geser kanan
hasil_2016 = angka1_2016 >> jumlah_geser_2016
print("\nBitwise Geser Kanan (>>)")
print(angka1_2016, ">>", jumlah_geser_2016, "=", hasil_2016)
print("Biner hasil =", bin(hasil_2016))
print("Biner hasil (8 bit) =", format(hasil_2016, '08b'))
```

## Tugas Praktikum

**Soal :**

 Studi Kasus: Sistem Simulasi Transaksi dan Validasi Akses Toko
Anda diminta untuk membuat sebuah program kasir sederhana yang digunakan untuk menghitung total transaksi pelanggan sekaligus melakukan pemeriksaan terhadap status pelanggan dan hak akses terhadap promo tertentu. Program menerima data transaksi dan informasi pelanggan, kemudian menggunakan berbagai jenis operator Python untuk menentukan total pembayaran, status diskon, kelayakan promo, serta hak akses pelanggan. Seluruh proses harus dibuat dalam satu alur program yang utuh.

**Kode Jawaban :**

a. Kode Program "tugas3_2611532016"

```python
print("=== SISTEM TRANSAKSI TOKO ===")

nama_2016 = input("Masukkan Nama Pelanggan : ")
status_2016 = input("Masukkan Status Pelanggan (member/nonmember) : ").lower()
total_belanja_2016 = float(input("Masukkan Total Belanja : Rp"))
jumlah_barang_2016 = int(input("Masukkan Jumlah Barang : "))
kode_promo_2016 = input("Masukkan Kode Promo : ").upper()

syarat_belanja_2016 = total_belanja_2016 >= 200000
syarat_barang_2016 = jumlah_barang_2016 >= 3
status_member_2016 = status_2016 == "member"

daftar_promo_2016 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

promo_tersedia_2016 = kode_promo_2016 in daftar_promo_2016
promo_tidak_tersedia_2016 = kode_promo_2016 not in daftar_promo_2016

diskon_member_2016 = status_member_2016 and syarat_belanja_2016

promo_didapatkan_2016 = promo_tersedia_2016 and (syarat_belanja_2016 or syarat_barang_2016)

bukan_member_2016 = not status_member_2016

if diskon_member_2016:
    persentase_diskon_2016 = 0.10
else:
    persentase_diskon_2016 = 0.05 if syarat_belanja_2016 else 0

besar_diskon_2016 = total_belanja_2016 * persentase_diskon_2016

total_pembayaran_2016 = total_belanja_2016 - besar_diskon_2016

if jumlah_barang_2016 > 0:
    rata_rata_barang_2016 = total_belanja_2016 / jumlah_barang_2016
else:
    rata_rata_barang_2016 = 0

sisa_pembagian_2016 = int(total_belanja_2016) % jumlah_barang_2016 if jumlah_barang_2016 > 0 else 0


poin_2016 = 0
if status_member_2016:
    poin_2016 += int(total_pembayaran_2016 // 10000)

jumlah_barang_tersisa_2016 = jumlah_barang_2016
if promo_didapatkan_2016:
    jumlah_barang_tersisa_2016 -= 1

kode_1_2016 = ["HEMAT10"]
kode_2_2016 = ["HEMAT10"]

nilai_sama_2016 = kode_1_2016 == kode_2_2016
objek_sama_2016 = kode_1_2016 is kode_2_2016
objek_berbeda_2016 = kode_1_2016 is not kode_2_2016


kode_member_2016 = 1 if status_member_2016 else 0
kode_belanja_2016 = 2 if syarat_belanja_2016 else 0
kode_barang_2016 = 4 if syarat_barang_2016 else 0
kode_promo_2016 = 8 if promo_tersedia_2016 else 0

# Operator OR (|)
kode_status_2016 = (
    kode_member_2016
    | kode_belanja_2016
    | kode_barang_2016
    | kode_promo_2016
)

cek_member_2016 = kode_status_2016 & 1
cek_belanja_2016 = kode_status_2016 & 2
cek_barang_2016 = kode_status_2016 & 4
cek_promo_2016 = kode_status_2016 & 8

kode_referensi_2016 = 11
perbandingan_status_2016 = kode_status_2016 ^ kode_referensi_2016

kode_shift_2016 = kode_status_2016 << 1


member_access_2016 = bool(cek_member_2016)
promo_access_2016 = bool(cek_promo_2016)
free_shipping_access_2016 = syarat_belanja_2016 and syarat_barang_2016


print("\n=== DATA PELANGGAN ===")
print("Nama Pelanggan       :", nama_2016)
print("Status Pelanggan     :", status_2016)
print("Total Belanja        : Rp", total_belanja_2016)
print("Jumlah Barang        :", jumlah_barang_2016)
print("Kode Promo           :", kode_promo_2016)


print("\n=== HASIL VALIDASI ===")
print("Belanja >= Rp200000  :", syarat_belanja_2016)
print("Jumlah Barang >= 3   :", syarat_barang_2016)
print("Status Member        :", status_member_2016)
print("Kode Promo Tersedia  :", promo_tersedia_2016)
print("Mendapatkan Diskon   :", diskon_member_2016)
print("Mendapatkan Promo    :", promo_didapatkan_2016)


print("\n=== HASIL PERHITUNGAN ===")
print("Besarnya Diskon      : Rp", besar_diskon_2016)
print("Total Pembayaran     : Rp", total_pembayaran_2016)
print("Rata-rata Harga      : Rp", rata_rata_barang_2016)
print("Sisa Pembagian       :", sisa_pembagian_2016)


print("\n=== HAK AKSES PELANGGAN ===")
print("Kode Hak Akses       :", format(kode_status_2016, "04b"))
print("Member Access        :", member_access_2016)
print("Promo Access         :", promo_access_2016)
print("Free Shipping Access :", free_shipping_access_2016)
print("Poin Pelanggan       :", poin_2016)


print("\n=== OPERATOR IDENTITAS ===")
print("kode_1 == kode_2     :", nilai_sama_2016)
print("kode_1 is kode_2     :", objek_sama_2016)
print("kode_1 is not kode_2 :", objek_berbeda_2016)


print("\n=== OPERASI BITWISE ===")
print("Kode Status Transaksi")
print("0001 | 0010 | 0100 | 1000")

print("Kode Biner           :", format(kode_status_2016, "04b"))
print("Kode Desimal         :", kode_status_2016)

print("\n=== PEMERIKSAAN STATUS ===")

print("\nCek Member")
print(format(kode_status_2016, "04b"), "& 0001")
print("Hasil Biner         :", format(cek_member_2016, "04b"))
print("Hasil Desimal       :", cek_member_2016)

print("\nCek Belanja")
print(format(kode_status_2016, "04b"), "& 0010")
print("Hasil Biner         :", format(cek_belanja_2016, "04b"))
print("Hasil Desimal       :", cek_belanja_2016)

print("\nCek Jumlah Barang")
print(format(kode_status_2016, "04b"), "& 0100")
print("Hasil Biner         :", format(cek_barang_2016, "04b"))
print("Hasil Desimal       :", cek_barang_2016)

print("\nCek Promo")
print(format(kode_status_2016, "04b"), "& 1000")
print("Hasil Biner         :", format(cek_promo_2016, "04b"))
print("Hasil Desimal       :", cek_promo_2016)

print("\n=== PERBANDINGAN STATUS (XOR) ===")
print("Kode Transaksi      :", format(kode_status_2016, "04b"))
print("Kode Referensi      :", format(kode_referensi_2016, "04b"))
print(format(kode_status_2016, "04b"), "^",
      format(kode_referensi_2016, "04b"))
print("Hasil Biner         :", format(perbandingan_status_2016, "04b"))
print("Hasil Desimal       :", perbandingan_status_2016)

print("\n=== SHIFT ===")
print(format(kode_status_2016, "04b"), "<< 1")
print("Hasil Biner         :", format(kode_shift_2016, "b"))
print("Hasil Desimal       :", kode_shift_2016)

print("\n=== SELESAI ===")
```
