## 2.1 Dasar Teori

#### 2.1.1 Struktur Percabangan

Struktur percabangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk membuat program mengambil keputusan berdasarkan kondisi tertentu. Dengan percabangan, program dapat menjalankan perintah yang berbeda sesuai dengan keadaan atau nilai yang diberikan oleh pengguna. Dalam Python, struktur percabangan yang umum digunakan antara lain `if`, `if-else`, `if-elif-else`, multi-if, dan `match-case`.

#### 2.1.2 Pernyataan if

Pernyataan `if` digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar (`True`). Jika kondisi bernilai salah (`False`), perintah di dalam `if` tidak akan dijalankan dan program akan melanjutkan ke perintah berikutnya.

Bentuk umum penggunaan `if` adalah:

`if kondisi:`

`    perintah`

Pada contoh program IPK, `if` digunakan untuk memeriksa apakah nilai IPK memenuhi batas yang telah ditentukan. Struktur ini cocok digunakan ketika hanya terdapat satu kondisi yang perlu diperiksa.

#### 2.1.3 Pernyataan if-else

Pernyataan `if-else` digunakan ketika terdapat dua kemungkinan hasil. Jika kondisi pada `if` bernilai benar, perintah pada bagian `if` akan dijalankan. Sebaliknya, jika kondisi bernilai salah, perintah pada bagian `else` akan dijalankan.

Bentuk umum:

`if kondisi:`

`    perintah jika benar`

`else:`

`    perintah jika salah`

Pada contoh program IPK, `if-else` digunakan untuk menentukan apakah pengguna lulus atau tidak berdasarkan nilai IPK.

#### 2.1.4 Pernyataan if-elif-else

Pernyataan `if-elif-else` digunakan ketika program memiliki lebih dari dua kondisi yang harus diperiksa. Program akan memeriksa kondisi secara berurutan mulai dari `if`, kemudian `elif`, dan terakhir `else`. Ketika menemukan kondisi yang bernilai benar, perintah pada kondisi tersebut dijalankan dan kondisi berikutnya tidak diperiksa.

Bentuk umum:

`if kondisi_1:`

`    perintah_1`

`elif kondisi_2:`

`    perintah_2`

`else:`

`    perintah_3`

Pada program pemeriksaan umur dan SIM C, struktur ini digunakan untuk menentukan apakah pengguna boleh membawa motor berdasarkan umur dan kepemilikan SIM. Struktur yang sama juga digunakan pada program konversi nilai angka menjadi nilai huruf.

#### 2.1.5 Multi-If

Multi-if adalah penggunaan beberapa pernyataan `if` secara terpisah dalam satu program. Berbeda dengan `if-elif-else`, setiap kondisi pada multi-if diperiksa secara independen. Artinya, lebih dari satu kondisi dapat bernilai benar dan semua perintah dari kondisi yang terpenuhi dapat dijalankan.

Multi-if dapat digunakan ketika beberapa kondisi dapat berlaku secara bersamaan. Contohnya terdapat pada program perhitungan diskon, yaitu diskon belanja besar, diskon member, dan diskon kode promo dapat diperoleh secara bersamaan. Setiap diskon diperiksa menggunakan `if` yang terpisah sehingga persentase diskon dapat diakumulasikan.

#### 2.1.6 Operator Perbandingan dan Operator Logika

Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan nilai `True` atau `False`. Beberapa operator yang digunakan pada program antara lain:

* `>` digunakan untuk memeriksa apakah suatu nilai lebih besar dari nilai lainnya.
* `>=` digunakan untuk memeriksa apakah suatu nilai lebih besar atau sama dengan nilai lainnya.
* `<` digunakan untuk memeriksa apakah suatu nilai lebih kecil dari nilai lainnya.
* `!=` digunakan untuk memeriksa apakah dua nilai tidak sama.
* `==` digunakan untuk memeriksa apakah dua nilai sama.

Selain operator perbandingan, terdapat operator logika yang digunakan untuk menggabungkan beberapa kondisi. Salah satunya adalah `and`, yang menghasilkan `True` jika seluruh kondisi yang digabungkan bernilai benar.

Contohnya pada pemeriksaan umur dan SIM C:

`umur_2016 >= 17 and sim_2016 == 'y'`

Kondisi tersebut hanya bernilai benar apabila umur pengguna minimal 17 tahun dan pengguna memiliki SIM C.

#### 2.1.7 Input dan Konversi Tipe Data

Fungsi `input()` digunakan untuk menerima data yang dimasukkan oleh pengguna melalui keyboard. Secara default, hasil dari `input()` berupa data bertipe string.

Untuk mendapatkan tipe data tertentu, input dapat dikonversi menggunakan fungsi seperti `int()` dan `float()`. Fungsi `int()` digunakan untuk mengubah data menjadi bilangan bulat, sedangkan `float()` digunakan untuk mengubah data menjadi bilangan desimal.

Contohnya:

`umur_2016 = int(input("Input umur anda: "))`

Pada kode tersebut, input umur diubah menjadi tipe data integer agar dapat digunakan dalam operasi perbandingan.

#### 2.1.8 Penggunaan strip(), lower(), dan [0]

Metode `.strip()` digunakan untuk menghapus spasi yang terdapat di awal atau akhir teks. Metode `.lower()` digunakan untuk mengubah seluruh huruf dalam teks menjadi huruf kecil. Kedua metode tersebut dapat digunakan untuk membuat input pengguna lebih mudah dibandingkan dengan kondisi yang telah ditentukan.

Contohnya:

`input_member_2016 = input("Apakah Anda member? (y/t): ").strip().lower()`

Selain itu, `[0]` dapat digunakan untuk mengambil karakter pertama dari sebuah string. Pada program pemeriksaan SIM C, `[0]` digunakan untuk mengambil karakter pertama dari jawaban pengguna, misalnya `y` atau `t`.

#### 2.1.9 Operator in

Operator `in` digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah kumpulan data, seperti list. Operator ini menghasilkan nilai `True` jika nilai ditemukan dan `False` jika tidak ditemukan.

Contohnya:

`input_member_2016 in ["y", "ya"]`

Kondisi tersebut digunakan untuk memeriksa apakah pengguna memasukkan jawaban `y` atau `ya`.

#### 2.1.10 match-case

`match-case` merupakan struktur percabangan pada Python yang digunakan untuk mencocokkan suatu nilai dengan beberapa pola atau pilihan yang tersedia. Struktur ini dapat digunakan sebagai alternatif ketika terdapat banyak pilihan yang berbeda.

Bentuk umum:

`match variabel:`

`    case nilai_1:`

`        perintah`

`    case nilai_2:`

`        perintah`

`    case _:`

`        perintah_default`

Pada program konversi angka menjadi nama bulan, `match-case` digunakan untuk mencocokkan angka 1 sampai 12 dengan nama bulan yang sesuai. `case _` digunakan sebagai kondisi default untuk menangani angka yang tidak sesuai dengan pilihan yang tersedia.

#### 2.1.11 Operasi Perhitungan Diskon

Dalam program perhitungan diskon, operator aritmatika digunakan untuk menghitung nominal diskon dan total pembayaran. Persentase diskon diubah terlebih dahulu menjadi bentuk desimal dengan membaginya dengan 100.

Rumus yang digunakan adalah:

Nominal Diskon = Total Belanja × (Total Diskon / 100)

Total Bayar = Total Belanja - Nominal Diskon

Pada program multi-if, beberapa persentase diskon dapat dijumlahkan terlebih dahulu apabila pengguna memenuhi beberapa syarat diskon.

#### 2.1.12 Konversi Nilai Angka Menjadi Nilai Huruf

Struktur `if-elif-else` dapat digunakan untuk mengelompokkan nilai angka ke dalam kategori tertentu. Pada program yang dibuat, nilai angka dikonversi menjadi nilai huruf berdasarkan rentang nilai.

Nilai 81 atau lebih dikategorikan sebagai A, nilai 70 sampai 80 sebagai B, nilai 60 sampai 69 sebagai C, nilai 50 sampai 59 sebagai D, dan nilai di bawah 50 sebagai E.

Penggunaan kondisi secara berurutan memungkinkan program menentukan satu kategori yang sesuai dengan nilai yang dimasukkan pengguna.

#### 2.1.13 Kesimpulan Dasar Teori

Berdasarkan teori tersebut, struktur percabangan merupakan bagian penting dalam pemrograman karena memungkinkan program mengambil keputusan berdasarkan kondisi tertentu. `if` digunakan untuk satu kondisi, `if-else` digunakan untuk dua kemungkinan, sedangkan `if-elif-else` digunakan untuk beberapa kondisi yang saling berurutan. Multi-if digunakan ketika beberapa kondisi perlu diperiksa secara independen dan dapat berlaku secara bersamaan. Sementara itu, `match-case` digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan yang tersedia.

Penggunaan operator perbandingan, operator logika, fungsi `input()`, serta konversi tipe data mendukung proses pengambilan keputusan dalam program. Konsep-konsep tersebut dapat diterapkan pada berbagai permasalahan seperti pemeriksaan kelayakan, penentuan nilai, perhitungan diskon, dan konversi angka menjadi nama bulan.

### 2.2.3 Program "multi_if1_2611532016"

```python
# Program ini menggunakan fungsi input()

umur_2016 = int(input("Masukkan umur Anda: "))
sim_2016 = input("Apakah Anda memiliki SIM c (y/t): ")[0]

if umur_2016 >= 17 and sim_2016 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_2016 >= 17 and sim_2016 != 'y':
    print("Anda sudah dewasa tapi tidak boleh bawa motor")

if umur_2016 < 17 and sim_2016 != 'y':
    print("Anda belum cukup umur bawa motor")

if umur_2016 < 17 and sim_2016 == 'y':
    print("Anda belum cukup umur punya SIM")
```

6. Baris 12 berfungsi untuk memeriksa apakah pengguna belum berusia 17 tahun dan tidak memiliki SIM C. Jika kedua kondisi terpenuhi, program menampilkan bahwa pengguna belum cukup umur untuk membawa motor.

### 2.2.4 Program "multi_if2_2611532016"

```python
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2016 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2016 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member = input_member_2016 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2016 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2016 = input_promo_2016 in ["y", "ya"]

total_diskon_persen_2016 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2016 > 1000000:
    total_diskon_persen_2016 += 10  # Diskon belanja besar

if is_member:
    total_diskon_persen_2016 += 5  # Diskon member

if kode_promo_valid_2016:
    total_diskon_persen_2016 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2016 = total_belanja_2016 * (total_diskon_persen_2016 / 100)
total_bayar_2016 = total_belanja_2016 - nominal_diskon_2016

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_2016}% (Rp {nominal_diskon_2016:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_2016:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2016}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan :**
1. Baris 1 merupakan komentar yang menjelaskan bahwa program menggunakan fungsi `input()` untuk menerima data dari pengguna.

2. Baris 2 merupakan komentar yang menjelaskan bahwa program digunakan untuk menghitung diskon belanja.

3. Baris 5 berfungsi untuk meminta pengguna memasukkan total belanja. Fungsi `float()` digunakan agar nilai yang dimasukkan dapat berupa bilangan desimal dan disimpan dalam variabel `total_belanja_2016`.

4. Baris 8 berfungsi untuk meminta status keanggotaan pengguna. Fungsi `.strip()` digunakan untuk menghapus spasi yang tidak diperlukan, sedangkan `.lower()` digunakan untuk mengubah input menjadi huruf kecil.

5. Baris 9 berfungsi untuk menentukan apakah pengguna merupakan member. Kondisi `in ["y", "ya"]` akan menghasilkan nilai `True` jika pengguna memasukkan `y` atau `ya`, kemudian hasilnya disimpan dalam variabel `is_member`.

6. Baris 12 berfungsi untuk meminta pengguna memasukkan status kode promo. Input dibersihkan menggunakan `.strip()` dan diubah menjadi huruf kecil menggunakan `.lower()`.

7. Baris 13 berfungsi untuk menentukan apakah kode promo yang dimasukkan valid. Jika pengguna memasukkan `y` atau `ya`, variabel `kode_promo_valid_2016` akan bernilai `True`.

8. Baris 15 berfungsi untuk menginisialisasi nilai total diskon sebesar `0%`. Nilai ini nantinya akan bertambah sesuai dengan kondisi yang terpenuhi.

9. Baris 20 berfungsi untuk memeriksa apakah total belanja lebih dari Rp1.000.000. Jika kondisi terpenuhi, program menambahkan diskon sebesar `10%` menggunakan operator `+=`.

10. Baris 23 berfungsi untuk memeriksa apakah pengguna merupakan member. Jika bernilai `True`, program menambahkan diskon member sebesar `5%`.

11. Baris 26 berfungsi untuk memeriksa apakah kode promo valid. Jika bernilai `True`, program menambahkan diskon voucher sebesar `15%`.

12. Baris 29 berfungsi untuk menghitung nominal diskon. Nilai total belanja dikalikan dengan persentase diskon yang telah dibagi `100` agar menjadi nilai desimal.

13. Baris 30 berfungsi untuk menghitung total pembayaran setelah dikurangi nominal diskon dari total belanja.

### 2.2.5 Program "if_elif_else1_2611532016"

```python
# Program ini menggunakan fungsi input()

umur_2016 = int(input("Input umur anda: "))
sim_2016 = input("Apakah Anda Sudah Punya Sim C: ")[0]

if umur_2016 >= 17 and sim_2016 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")
elif umur_2016 >= 17 and sim_2016 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")
elif umur_2016 < 17 and sim_2016 == 'y':
    print("Anda Belum Cukup Umur punya SIM")
else:
    print("Anda Belum Cukup Umur dan tidak boleh bawa motor")

print("Program Selesai")
```

### 2.2.7 Program "match_case1_2611532016"

```python
# Program konversi angka menjadi nama bulan

bulan_2016 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_2016:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

3. Baris 5 berfungsi untuk memulai struktur percabangan `match-case`. Nilai pada variabel `bulan_2016` akan dibandingkan dengan setiap `case` yang tersedia.

4. Baris 6–7 berfungsi untuk memeriksa apakah angka yang dimasukkan adalah 1. Jika sesuai, program menampilkan nama bulan "Januari".

5. Baris 8–9 berfungsi untuk memeriksa angka 2 dan menampilkan nama bulan "Februari" jika sesuai.

6. Baris 10–11 berfungsi untuk memeriksa angka 3 dan menampilkan nama bulan "Maret" jika sesuai.

7. Baris 12–13 berfungsi untuk memeriksa angka 4 dan menampilkan nama bulan "April" jika sesuai.

8. Baris 14–15 berfungsi untuk memeriksa angka 5 dan menampilkan nama bulan "Mei" jika sesuai.

9. Baris 16–17 berfungsi untuk memeriksa angka 6 dan menampilkan nama bulan "Juni" jika sesuai.

10. Baris 18–19 berfungsi untuk memeriksa angka 7 dan menampilkan nama bulan "Juli" jika sesuai.

11. Baris 20–21 berfungsi untuk memeriksa angka 8 dan menampilkan nama bulan "Agustus" jika sesuai.

## Tugas Praktikum

**Soal :**

Sebuah wahana rekreasi petualangan membutuhkan sistem loket berbasis konsol untuk menentukan jenis wahana, memverifikasi kelaikan pengunjung dengan kondisi majemuk (logical operators), menghitung diskon akumulatif menggunakan multi-if terpisah, serta mencetak audit status operasional.

**Kode Jawaban :**

a. Kode Program "Tugas4__2611532016"

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. INPUT DATA PENGUNJUNG

nama_pengunjung_2016 = input("Masukkan Nama Pengunjung        : ")
umur_2016 = int(input("Input umur anda                 : "))
sim_2016 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()
jumlah_tiket_2016 = int(input("Masukkan jumlah tiket           : "))

# Validasi jumlah tiket menggunakan if tunggal
if jumlah_tiket_2016 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

# 2. PEMILIHAN PAKET WAHANA MENGGUNAKAN MATCH-CASE

print("\nPilihan Paket Wahana (1-5):")
print(" 1. Safari Rimba         (Rp 50,000)")
print(" 2. Arung Jeram          (Rp 75,000)")
print(" 3. Motor ATV Ekstrim    (Rp 120,000)")
print(" 4. Roller Coaster Kilat (Rp 100,000)")
print(" 5. All-Access VIP       (Rp 220,000)")

paket_2016 = int(input("Masukkan nomor paket (1-5) : "))

match paket_2016:
    case 1:
        nama_wahana_2016 = "Wahana Safari Rimba"
        harga_satuan_2016 = 50000

    case 2:
        nama_wahana_2016 = "Wahana Arung Jeram"
        harga_satuan_2016 = 75000

    case 3:
        nama_wahana_2016 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2016 = 120000

    case 4:
        nama_wahana_2016 = "Wahana Roller Coaster Kilat"
        harga_satuan_2016 = 100000

    case 5:
        nama_wahana_2016 = "Wahana All-Access VIP"
        harga_satuan_2016 = 220000

    case _:
        print("Paket wahana tidak valid!")
        raise SystemExit


# 3. VALIDASI IZIN KENDALI WAHANA
#    Menggunakan if-elif-else dan operator logika

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_2016 == 3 and umur_2016 >= 17 and sim_2016 == 'y':
    status_akses_2016 = (
        "Anda sudah dewasa dan boleh mengendarai ATV sendiri."
    )

elif paket_2016 == 3 and umur_2016 >= 17 and sim_2016 != 'y':
    status_akses_2016 = (
        "Anda sudah dewasa tetapi tidak boleh bawa motor ATV "
        "(wajib didampingi instruktur)."
    )

elif paket_2016 == 3 and umur_2016 < 17 and sim_2016 == 'y':
    status_akses_2016 = (
        "Identitas tidak valid: Belum cukup umur memiliki SIM."
    )

elif paket_2016 == 3:
    status_akses_2016 = (
        "Anda belum cukup umur dan tidak boleh bawa motor ATV."
    )

elif umur_2016 >= 10:
    status_akses_2016 = (
        "Anda memenuhi syarat umur untuk wahana ini."
    )

else:
    status_akses_2016 = (
        "Anda belum cukup umur untuk wahana ini."
    )

print(f"Status Akses: {status_akses_2016}")


# 4. INPUT DATA DISKON

is_member_2016 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_2016 = input(
    "Apakah kode promo valid? (y/t) : "
).strip().lower()


# 5. PERHITUNGAN SUBTOTAL

subtotal_2016 = harga_satuan_2016 * jumlah_tiket_2016


# 6. MULTI-IF UNTUK DISKON AKUMULATIF

total_diskon_persen_2016 = 0

# Diskon Belanja Besar
if subtotal_2016 >= 200000:
    total_diskon_persen_2016 += 10

# Diskon Member
if is_member_2016 in ['y', 'ya']:
    total_diskon_persen_2016 += 5

# Diskon Voucher Promo
if kode_promo_valid_2016 in ['y', 'ya']:
    total_diskon_persen_2016 += 15

# Diskon Tambahan Rombongan
if jumlah_tiket_2016 >= 5:
    total_diskon_persen_2016 += 5


# 7. PERHITUNGAN NOMINAL DISKON DAN TOTAL BAYAR

nominal_diskon_2016 = (
    subtotal_2016 * (total_diskon_persen_2016 / 100)
)

total_bayar_2016 = subtotal_2016 - nominal_diskon_2016


# 8. EVALUASI KELULUSAN AUDIT

if total_bayar_2016 > 300000:
    catatan_layanan_2016 = (
        "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    )
else:
    catatan_layanan_2016 = (
        "Terima kasih telah berkunjung."
    )


# 9. RINCIAN PEMBAYARAN

print("\n--- Rincian Pembayaran ---")
print(f"Nama Pengunjung  : {nama_pengunjung_2016}")
print(f"Wahana           : {nama_wahana_2016}")
print(f"Harga Satuan     : Rp {harga_satuan_2016:,.0f}")
print(f"Jumlah Tiket     : {jumlah_tiket_2016}")
print(f"Subtotal Belanja : Rp {subtotal_2016:,.0f}")
print(
    f"Total Diskon     : {total_diskon_persen_2016}% "
    f"(Rp {nominal_diskon_2016:,.0f})"
)
print(f"Total Bayar      : Rp {total_bayar_2016:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_2016}")

print("\nProgram Selesai")
```

**Baris 8–9** berfungsi untuk melakukan validasi jumlah tiket menggunakan `if` tunggal. Jika jumlah tiket kurang dari atau sama dengan 0, program akan menampilkan peringatan bahwa kuota tiket tidak valid.

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa struktur percabangan merupakan bagian penting dalam pemrograman Python karena digunakan untuk mengatur alur program berdasarkan kondisi tertentu. Dengan struktur percabangan, program dapat menentukan tindakan yang berbeda sesuai dengan data atau kondisi yang diberikan oleh pengguna.

Dalam praktikum ini dipelajari beberapa bentuk struktur percabangan, yaitu if, if-else, if-elif-else, multi-if, dan match-case. Masing-masing memiliki cara penggunaan yang berbeda sesuai dengan kebutuhan program. if digunakan untuk satu kondisi, if-else digunakan untuk dua kemungkinan, sedangkan if-elif-else digunakan untuk menangani beberapa kondisi secara berurutan. Multi-if digunakan untuk memeriksa beberapa kondisi secara independen sehingga lebih dari satu kondisi dapat dijalankan. Sementara itu, match-case digunakan untuk memilih tindakan berdasarkan nilai yang sesuai dengan beberapa pilihan yang tersedia.

Penggunaan operator logika seperti and, or, dan != juga dapat membantu membentuk kondisi yang lebih kompleks.
